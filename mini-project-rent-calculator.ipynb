```python
# Simple Mini project in python (Rent Calculator)
'''total rent
total food order for snacking
total electricity units spend
charge per unit
persons living in Room'''

# output: total ammount you've to pay is:

# Input from User
rent=int(input("Entre flat Rent = "))
food=int(input("Enter the amount of food order = "))
electricity_spend=int(input("Enter the total of electricity spend = "))
charge_per_unit=int(input("Enter the Charge per unit = "))
Persons=int(input("Enter the number of persons living in flat = "))

# Operations

total_electricity_bill = electricity_spend * charge_per_unit

total_bill=(rent+food+total_electricity_bill)/Persons

# printing Output Sentence
print("Each Persoon will pay = ",total_bill)
```
