```python
# ☕ Café Management System
# ==========================================
#       ☕ CAFE MANAGEMENT SYSTEM ☕
# ==========================================

# Cafe Menu
menu = {
    "Coffee": 50,
    "Tea": 30,
    "Sandwich": 80,
    "Burger": 100,
    "Pizza": 150
}

# Store customer order
order = {}


# Function to display menu
def display_menu():
    print("\n========== CAFE MENU ==========")

    for item, price in menu.items():
        print(item, ": ₹", price)

    print("===============================")


# Function to take order
def take_order():

    display_menu()

    while True:

        item = input("\nEnter item name / (type 'done' to finish): ")

        # Stop ordering
        if item.lower() == "done":
            break

        # Check whether item exists
        if item in menu:

            quantity = int(input("Enter quantity: "))

            # Add item to order
            if item in order:
                order[item] = order[item] + quantity
            else:
                order[item] = quantity

            print(quantity, item, "added to your order.")

        else:
            print("Sorry, this item is not available.")

    print("\nOrder completed!")


# Function to generate bill
def generate_bill():

    if len(order) == 0:
        print("\nNo items ordered.")
        return

    print("\n========== BILL ==========")

    total = 0

    for item, quantity in order.items():

        price = menu[item]
        item_total = price * quantity

        print(item, "x", quantity, "=", "₹", item_total)

        total = total + item_total

    print("--------------------------")
    print("Total Amount: ₹", total)
    print("==========================")
    print("Thank you for visiting!")
    print("Please visit again ☕")


# Main program
while True:

    print("\n================================")
    print("       ☕ MY CAFE ☕")
    print("================================")

    print("1. Display Menu")
    print("2. Order Food")
    print("3. Generate Bill")
    print("4. Exit")

    choice = input("\nEnter your choice: ")

    if choice == "1":

        display_menu()

    elif choice == "2":

        take_order()

    elif choice == "3":

        generate_bill()

    elif choice == "4":

        print("\nThank you for using Cafe Management System!")
        break

    else:

        print("\nInvalid choice. Please try again.")
```
